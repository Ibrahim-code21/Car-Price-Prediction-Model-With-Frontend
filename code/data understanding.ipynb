```python
import pandas as pd
data = pd.read_csv("../files/car_data.csv")

shape = data.shape

columns = data.columns


describe = data.describe()

null_values = data.isnull().sum()

data_types = data.dtypes

catagorical_columns = data.select_dtypes(include=['object']).columns

numerical_columns = data.select_dtypes(exclude=['object']).columns

max_car_age = data["Car_Age"].max()

max_km = data["Driven_kms"].max()

print("Shape of the dataset: \n", shape)
print("-------------------------------")
print("Columns in the dataset: \n", columns)
print("-------------------------------")
print("description of the dataset: \n", describe)
print("-------------------------------")
print("Null values in each column: \n", null_values)
print("-------------------------------")
print("Data types of each column:\n", data_types)
print("-------------------------------")
print("Categorical columns: \n", catagorical_columns)
print("-------------------------------")
print("Numerical columns: \n", numerical_columns)
print("-------------------------------")
print("max Car Age is: \n", max_car_age)
print("-------------------------------")
print("max Car Age is: ",max_km)



```
